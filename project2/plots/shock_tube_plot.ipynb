```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt('results_100.dat')
xa = data[:,0]
xb = data[:,1]
rho = data[:,2]
v = data[:,3]
e_d = data[:,4]
P = data[:,5]

fig, ax = plt.subplots(2, 2, figsize=(10,8))
axs = ax.flatten()
axs[0].scatter(xa, rho, label='Density (d)',c='C0',s=5)
axs[1].scatter(xa, P, label='Pressure (p)',c='C1',s=5)
axs[2].scatter(xa, v, label='Velocity (v)',c='C2',s=5)
axs[3].scatter(xa, e_d, label='Specific Energy (e/d)',c='C3',s=5)

for i in range(0,4):
    axs[i].set_xlabel('x')
    axs[i].set_ylabel('value')
    axs[i].legend()
    axs[i].set_ylim(min(axs[i].get_ylim()),max(axs[i].get_ylim())+0.05*(max(axs[i].get_ylim())-min(axs[i].get_ylim())))
plt.suptitle('Test Cartesian shock tube problem at t=0.245 with 100 grid points')
fig.tight_layout(rect=[0, 0, 1, 1.0])
plt.savefig('CST100.pdf', format='pdf',dpi=300,bbox_inches='tight')
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

data = np.loadtxt('SST_100.dat')
xa = data[:,0]
xb = data[:,1]
rho = data[:,2]
v = data[:,3]
e_d = data[:,4]
P = data[:,5]
m = data[:,6]

fig = plt.figure(figsize=(12, 8))
gs = gridspec.GridSpec(2, 3, height_ratios=[1, 1.2])

# upper three plots (each takes 1 column)
ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[0, 1])
ax3 = fig.add_subplot(gs[0, 2])

# lower two plots (first takes 2 columns, second takes 1 column)
ax4 = fig.add_subplot(gs[1, 0])
ax5 = fig.add_subplot(gs[1, 1])

ax1.scatter(xa, rho, label='Density (d)', c='C0', s=5)
ax2.scatter(xa, P, label='Pressure (p)', c='C1', s=5)
ax3.scatter(xa, v, label='Velocity (v)', c='C2', s=5)
ax4.scatter(xa, e_d, label='Specific Energy (e/d)', c='C3', s=5)
ax5.scatter(xa, rho * v, label='Momentum (s)', c='C4', s=5)

for ax in [ax1,ax2,ax3,ax4,ax5]:
    ax.set_xlabel('x')
    ax.set_ylabel('value')
    ax.legend(loc='best')
    ax.set_ylim(min(ax.get_ylim()),max(ax.get_ylim())+0.05*(max(ax.get_ylim())-min(ax.get_ylim())))
plt.suptitle('Strong shock tube at t=2.3 with 100 grid points')
plt.tight_layout()
plt.savefig('Strong100.pdf', format='pdf',dpi=300,bbox_inches='tight')
```
